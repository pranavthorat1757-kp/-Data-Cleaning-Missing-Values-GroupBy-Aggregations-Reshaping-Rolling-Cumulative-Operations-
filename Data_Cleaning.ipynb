**Question 2:** Explain median imputation and mode imputation with suitable examples. Why is the median preferred for numerical columns such as age, and why is the mode used for categorical columns such as embarked?    tell me short ans .

**Median imputation** replaces missing numerical values with the median of that column.

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
```

**Mode imputation** replaces missing categorical values with the most frequent value.

```python
df["Embarked"].fillna(df["Embarked"].mode()[0], inplace=True)
```

**Why?**

* **Median** is preferred for numerical columns like `Age` because it is less affected by outliers and skewed data.
* **Mode** is used for categorical columns like `Embarked` because it represents the most common category.

**Question 5:** Explain the difference between agg(), transform() and filter() in a groupby
operation. Support your answer with one example of each.


In Pandas `groupby()`:

* **`agg()`** – Performs calculations on each group and returns summarized results.

```python
df.groupby("Department")["Salary"].agg("mean")
```

* **`transform()`** – Performs calculations on each group and returns a result with the **same size as the original data**.

```python
df["Avg_Salary"] = df.groupby("Department")["Salary"].transform("mean")
```

* **`filter()`** – Keeps or removes entire groups based on a condition.

```python
df.groupby("Department").filter(lambda x: x["Salary"].mean() > 30000)
```

**In short:** `agg()` → summary, `transform()` → same-size result, `filter()` → selects groups.


**Question 6:**  Use the following dataset:
import pandas as pd

df = pd.DataFrame({

'region': ['North','North','South','South','East','East','East'],

'category': ['A','B','A','A','B','A','B'],

'sales': [100, 150, 200, 250, 120, 90, 300],

'discount': [5, 10, 0, 2, 3, 1, 4],

'quantity': [10, 12, 15, 20, 8, 9, 14]
})

Write a Python program to:
1. Group the data by region and category and find the total sales of every group.
2. Find the number of records present in every group.
3. Create a region-wise summary containing total_sales, avg_discount and max_qty using
named aggregation.
4. Find the mean, min and max sales of every region–category pair.
```python
import pandas as pd

df = pd.DataFrame({
    'region': ['North','North','South','South','East','East','East'],
    'category': ['A','B','A','A','B','A','B'],
    'sales': [100,150,200,250,120,90,300],
    'discount': [5,10,0,2,3,1,4],
    'quantity': [10,12,15,20,8,9,14]
})

# 1. Total sales by region and category
print(df.groupby(['region','category'])['sales'].sum())

# 2. Number of records in every group
print(df.groupby(['region','category']).size())

# 3. Region-wise summary using named aggregation
summary = df.groupby('region').agg(
    total_sales=('sales', 'sum'),
    avg_discount=('discount', 'mean'),
    max_qty=('quantity', 'max')
)
print(summary)

# 4. Mean, minimum and maximum sales
result = df.groupby(['region','category'])['sales'].agg(
    ['mean', 'min', 'max']
)
print(result)
```

This program uses **`groupby()`**, **`sum()`**, **`size()`**, and **named aggregation** to analyze the dataset.

**Question 7:** Using the same dataset given in Question 6, write a Python program to:

1. Add a column region_avg_sales containing the average sales of the region of each row.

2. Add a column pct_contribution showing the share of each row in its region's total sales.

3. Add a column rank_region_sales that ranks the sales within each region in descending order.

4. Keep only those regions that contain more than two records.

```python
import pandas as pd

df = pd.DataFrame({
    'region': ['North','North','South','South','East','East','East'],
    'category': ['A','B','A','A','B','A','B'],
    'sales': [100,150,200,250,120,90,300],
    'discount': [5,10,0,2,3,1,4],
    'quantity': [10,12,15,20,8,9,14]
})

# 1. Average sales of each region
df['region_avg_sales'] = df.groupby('region')['sales'].transform('mean')

# 2. Percentage contribution to regional sales
df['pct_contribution'] = (
    df['sales'] / df.groupby('region')['sales'].transform('sum') * 100
)

# 3. Rank sales within each region
df['rank_region_sales'] = df.groupby('region')['sales'].rank(
    ascending=False
)

# 4. Keep regions with more than two records
df = df.groupby('region').filter(lambda x: len(x) > 2)

print(df)
```

**Key functions:** `transform()` calculates values for each row, `rank()` ranks sales within each region, and `filter()` keeps regions having more than two records.

'month': ['Jan','Feb','Jan','Feb','Jan','Feb'],

'sales': [100, 120, 200, 220, 150, 180],

'discount': [5, 7, 3, 4, 6, 8]
})
Write a Python program to:
1. Convert the data into a wide format using pivot(), with region as the index and the
months as columns.
2. Create a pivot table of total sales along with an overall Total row and column.
3. Create a pivot table of the average discount in which the missing combinations are filled
with 0.
4. Explain in which situation pivot() fails but pivot_table() works.

```python
import pandas as pd

df = pd.DataFrame({
    'region': ['North','North','South','South','East','East'],
    'month': ['Jan','Feb','Jan','Feb','Jan','Feb'],
    'sales': [100,120,200,220,150,180],
    'discount': [5,7,3,4,6,8]
})

# 1. Wide format using pivot()
wide = df.pivot(index='region', columns='month', values='sales')
print(wide)

# 2. Total sales with Total row and column
sales_pivot = pd.pivot_table(
    df, index='region', columns='month',
    values='sales', aggfunc='sum', margins=True
)
print(sales_pivot)

# 3. Average discount with missing values as 0
discount_pivot = pd.pivot_table(
    df, index='region', columns='month',
    values='discount', aggfunc='mean', fill_value=0
)
print(discount_pivot)
```

**Explanation:**
`pivot()` fails when there are **duplicate index-column combinations**. `pivot_table()` works in such cases because it can **aggregate duplicate values** using functions like `sum()`, `mean()`, etc.

**Question 10:** Use the following datasets:
import pandas as pd
import numpy as np
sales = pd.DataFrame({

'order_id': [1, 2, 3, 4, 5, 6],

'region': ['North','South','East','North','South','East'],

'category': ['A','B','A','B','A','B'],

'sales': [100, 250, np.nan, 160, 200, 300],

'quantity': [10, 20, 12, 10, 15, 14],

'rep_id': [101, 102, 103, 101, 102, 103]

})
reps = pd.DataFrame({
  
'rep_id': [101, 102, 103],

'rep_name': ['Aditi', 'Rahul', 'Sneha'],

'experience': [5, 2, 7]
})
Write a single Python program to:
1. Display the number of missing values in every column and fill the missing sales with the
median.
2. Merge sales with reps on rep_id using a left join.
3. Create a column revenue_per_unit as sales / quantity.
4. Create a region-wise summary containing total_sales, avg_sales and orders.
5. Create a pivot table of total sales with region as the index and category as the columns.
6. Print the name of the representative who generated the highest total sales.

```python
import pandas as pd
import numpy as np

sales = pd.DataFrame({
    'order_id': [1,2,3,4,5,6],
    'region': ['North','South','East','North','South','East'],
    'category': ['A','B','A','B','A','B'],
    'sales': [100,250,np.nan,160,200,300],
    'quantity': [10,20,12,10,15,14],
    'rep_id': [101,102,103,101,102,103]
})

reps = pd.DataFrame({
    'rep_id': [101,102,103],
    'rep_name': ['Aditi','Rahul','Sneha'],
    'experience': [5,2,7]
})

# 1. Missing values and median imputation
print(sales.isna().sum())
sales['sales'] = sales['sales'].fillna(sales['sales'].median())

# 2. Left join
df = sales.merge(reps, on='rep_id', how='left')

# 3. Revenue per unit
df['revenue_per_unit'] = df['sales'] / df['quantity']

# 4. Region-wise summary
summary = df.groupby('region').agg(
    total_sales=('sales', 'sum'),
    avg_sales=('sales', 'mean'),
    orders=('order_id', 'count')
)
print(summary)

# 5. Pivot table
pivot = pd.pivot_table(
    df, index='region', columns='category',
    values='sales', aggfunc='sum', fill_value=0
)
print(pivot)

# 6. Representative with highest total sales
top_rep = df.groupby('rep_name')['sales'].sum().idxmax()
print("Highest sales representative:", top_rep)
```

**Output:** The representative with the highest total sales is **Sneha**.
